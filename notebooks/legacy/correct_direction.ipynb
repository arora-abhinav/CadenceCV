```python
import json
import cv2
with open("/Users/abhinavarora/Desktop/CadenceCV/data/strikefoot_data.json", "r") as file:
    data = json.load(file)

video_direction = {}
videos = set()
for strike in data:
    videos.add(strike["video"])

base_video_dir = "/Users/abhinavarora/Desktop/CadenceCV/Videos"
video_extensions = [".mp4", ".MOV"]
for video in videos:
    for ext in video_extensions:
        try:
            dir = base_video_dir + "/" + video + ext
            cap = cv2.VideoCapture(dir)
            #Seeing the first frame of each video and checking if the runner is facing right or left
            cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
            ret, img = cap.read()
            cv2.imshow("Frame", img)
            key = cv2.waitKey(0)
            if key == ord("l"):
                video_direction[video] = 'L'
            else:
                video_direction[video] = "R"

            break
        except:
            continue

for strike in data:
    if video_direction[strike["video"]] == 'L':
        if strike["foot_angle"] != None:
            strike["foot_angle"] *= -1
        strike["ankle_hip_horizontal_offset"] *= -1
        strike["ankle_x_vel"] *= -1
        strike["tibial_angle"] *= -1
        strike["ankle_approach_angle"] *= -1
```

```python
with open("/Users/abhinavarora/Desktop/CadenceCV/data/strikefoot_data.json", "w") as file:
    json.dump(data, file, indent=4)
```
