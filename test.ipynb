```python
import cv2
from ultralytics import YOLO
import sys
import torch
sys.path.append("/Users/abhinavarora/Desktop/CadenceCV/src")
sys.path.append("/Users/abhinavarora/Desktop/CadenceCV/models")
from yolo_model_loader import load_model, load_source

model = load_model("/Users/abhinavarora/Desktop/CadenceCV/models/21_keypoints.pt")
cap = cv2.imread("Test Frame.png", cv2.IMREAD_COLOR)
source = load_source(cap, model)
for res in source:
    kpts = res.keypoints.xy[0]

for point in kpts:
    p:torch.tensor = point
    cap = cv2.circle(cap, tuple(p.int().tolist()), 2, (0,0,255), -1)

cv2.imshow("Image", cap)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
